# Embedding Techniques using HuggingFace

Hugging Face provides a wide range of pre-trained models that can convert text into numerical vector representations called embeddings.

Common uses of Hugging Face embeddings are:

- Semantic Search
- Document Similarity
- Question Answering
- Recommendation Systems
- RAG (Retrieval-Augmented Generation)
- Text Classification

## Sentence Transformers on Hugging Face

SentenceTransformer loads the pre-trained Hugging Face model. Hugging Face sentence-transformers is a Python framework for state-of-the-art sentence, text and image embeddings. One of the embedding models is used in the HuggingFaceEmbeddings class.

### Loading HuggingFace Embeddings

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

### Generating Query Embeddings

```python
text = "This is a test document"
query_result = embeddings.embed_query(text)
query_result
```

### Check embedding dimensions

```python
print(type(query_result))
len(query_result)
```

```python
print(query_result[:10])
```

This helps you understand that an embedding is simply a list of numbers with a fixed dimension.

### Generating Document Embeddings

```python
doc_result = embeddings.embed_documents([text, "This is not a test document"])
doc_result[0]
```

### Embed multiple documents

```python
documents = [
    "Python is a programming language.",
    "Java is also a programming language.",
    "I love eating pizza.",
    "Machine learning is a branch of artificial intelligence."
]

doc_embeddings = embeddings.embed_documents(documents)

print(len(doc_embeddings))
print(len(doc_embeddings[0]))
```

4 documents -> 4 embedding vectors -> each vector has the same dimensions

### Compare query and document embeddings

```python
from sklearn.metrics.pairwise import cosine_similarity

query = "I want to learn Python"

documents = [
    "Python is a programming language.",
    "Java is a programming language.",
    "I love eating pizza.",
    "Machine learning uses data."
]

query_embedding = embeddings.embed_query(query)
document_embeddings = embeddings.embed_documents(documents)

similarities = cosine_similarity(
    [query_embedding],
    document_embeddings
)

print(similarities)
```

The important idea is:
Query
"I want to learn Python"
        ->
Embedding
        ->
Compare with document embeddings
        ->
Find the most similar document

### Find the most similar document

```python
scores = similarities[0]

for document, score in zip(documents, scores):
    print(f"{score:.4f} -> {document}")
```

### Find the highest score

```python
best_index = scores.argmax()

print("Most similar document:")
print(documents[best_index])

print("Similarity score:")
print(scores[best_index])
```

This is essentially the basic idea behind semantic search.
