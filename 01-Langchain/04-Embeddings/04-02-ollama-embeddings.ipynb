# Embedding Techniques using Ollama

Ollama supports embedding models, making it possible to build retrieval augmented generation (RAG) applications that combine text prompts with existing documents or other data.

The main advantage is that the embedding model runs on your own computer, so you don't need to send the text to a cloud API.

### Import Ollama Embeddings

```python
from langchain_community.embeddings import OllamaEmbeddings
```

### Initialize Ollama Embedding Model

```python
embeddings = (
    OllamaEmbeddings(model="mxbai-embed-large")         # by default it uses llama2
)
```

```python
embeddings
```

### Generate Embeddings for Documents

```python
r1 = embeddings.embed_documents(
    [
        "Iron Man is better than Captain America",
        "LOKI is the God of Multiverse",
    ]
)
```

```python
r1
```

```python
r1[0]
```

### Check Embedding Dimension

```python
len(r1[0])
```

### Generate Query Embedding

```python
r2 = embeddings.embed_query("Who is better than Captain America")
print(r2)
```

Other Embedding models: https://ollama.com/blog/embedding-models

### Calculate Similarity Between Query and Documents

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(
    [r2],
    r1
)

similarity
```

### Print Similarity with Each Document

```python
for document, score in zip(r1, similarity[0]):
    print(f"Score: {score:.4f} → {document}")
```

### Find the Most Similar Document

```python
scores = similarity[0]

best_index = scores.argmax()

print("Most Similar Document:")
print(r1[best_index])

print("Similarity Score:")
print(scores[best_index])
```

### Try Multiple Queries

```python
queries = [
    "Who is better than Captain America?",
    "Who is the God of Multiverse?",
    "Tell me about Iron Man"
]

for query in queries:
    query_embedding = embeddings.embed_query(query)

    similarity = cosine_similarity(
        [query_embedding],
        r1
    )

    best_index = similarity[0].argmax()

    print("\nQuery:", query)
    print("Most Similar:", r1[best_index])
    print("Score:", similarity[0][best_index])
```

### Understand the Complete Flow

Documents
    ->
Ollama Embeddings
    ->
Document Vectors
    ->
User Query
    ->
Query Vector
    ->
Cosine Similarity
    ->
Most Similar Document
