# 🛡️ Introduction to Guardrails

Guardrails help you build **safe and compliant AI applications** by validating and filtering content at key points in an agent's execution.

In an AI agent, guardrails act as **middleware** that can intercept execution and control what enters, what happens during execution, and what leaves the system.

---

## 🧠 What are Guardrails?

A guardrail is a layer of protection placed around an AI system to make sure that the system behaves within defined **safety, security, privacy, and business rules**.

Guardrails can:

- Validate user inputs
- Detect unsafe or malicious requests
- Protect sensitive information
- Control access to tools
- Require human approval for sensitive actions
- Validate AI-generated outputs
- Enforce application-specific business rules

In LangChain, guardrails can be implemented using **middleware** that intercepts different stages of agent execution.

---

## 🔄 Where Guardrails Fit in an Agent

Guardrails can operate at different points during an agent's execution:

```text
                User Input
                    │
                    ▼
          ┌───────────────────┐
          │   Input Guardrail │
          │ Validate / Filter │
          └─────────┬─────────┘
                    │
                    ▼
             ┌─────────────┐
             │ AI Agent    │
             └──────┬──────┘
                    │
             ┌──────▼──────┐
             │ Model/Tools │
             └──────┬──────┘
                    │
                    ▼
          ┌───────────────────┐
          │  Output Guardrail │
          │ Validate / Filter │
          └─────────┬─────────┘
                    │
                    ▼
              Final Response

---

### Main Guardrail Positions

| Position | Purpose |
|---|---|
| **Before Agent** | Validate or filter user input before the agent starts |
| **Around Model/Tools** | Control model calls or tool execution |
| **After Agent** | Validate the final response before it reaches the user |

---

## 🎯 Why Do We Need Guardrails?

LLMs are powerful but they do not automatically follow every application rule.

An agent may:

- Receive malicious or adversarial input
- Accidentally expose sensitive information
- Generate unsafe content
- Call a sensitive tool without approval
- Produce an output that violates business rules
- Be manipulated through prompt injection
- Generate incorrect or inappropriate responses

Guardrails provide an additional **control layer** around the AI system.

---

## 🔐 Common Use Cases

| Use Case | Example |
|---|---|
| **PII leakage prevention** | Redact emails or credit card numbers |
| **Prompt injection blocking** | Detect adversarial instructions |
| **Harmful content filtering** | Block dangerous requests |
| **Business rule enforcement** | Require approval for financial operations |
| **Tool access control** | Prevent unauthorized tool execution |
| **Output validation** | Ensure responses meet safety requirements |
| **Compliance** | Enforce application-specific policies |
| **Human approval** | Require confirmation before sensitive actions |

---

## 🧩 Example

Imagine an AI agent that can send emails.

### Without a Guardrail

```text
User
  ↓
AI Agent
  ↓
send_email()
  ↓
Email Sent
```
The agent may execute the action automatically.

### With a Guardrail

```text
User
  ↓
AI Agent
  ↓
Request to send email
  ↓
Human Approval Guardrail
  ↓
Approved?
  ├── Yes → send_email()
  └── No  → Action blocked
```
The guardrail adds a safety checkpoint before the sensitive operation.

---

## 💡 Key Idea

> **Guardrails don't replace the AI agent. They control and protect the agent's execution.**

The agent is responsible for accomplishing the task.

The guardrails are responsible for ensuring that the task is performed within defined **safety, privacy, security, and business constraints**.

---

## 📝 Key Takeaways

- Guardrails are protection and validation layers around AI applications.
- They can validate **inputs and outputs**.
- They can control **model and tool execution**.
- LangChain implements guardrails through **middleware**.
- Guardrails can protect against **PII leakage, prompt injection, harmful content, and unauthorized actions**.
- Sensitive operations can require **human approval**.
- Multiple guardrails can be combined to create **layered protection**.
- Guardrails are especially important when AI agents can interact with **tools, databases, APIs, or real-world systems**.
