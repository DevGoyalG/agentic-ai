# 🛡️ Guardrails with LangChain — Summary

Guardrails help make AI agents **safer, more reliable, and better controlled** by validating inputs, protecting sensitive information, controlling tool execution, and validating outputs.

---

## 📚 Guardrail Types

| Guardrail Type | Hook | When it Runs | Best For |
|---|---|---|---|
| PII Middleware | Input / Output | Around agent execution | Data privacy & compliance |
| Human-in-the-Loop | Tool level | Before sensitive tools | High-stakes decisions |
| Content Filter | `before_agent` | Start of invocation | Blocking bad inputs early |
| Safety Validator | `after_agent` | End of invocation | Output safety |
| Custom Logic | Any hook | Wherever required | Business rules |

---

## 🔑 Key Takeaways

1. **Guardrails = Middleware** — use them with `middleware=[]` in `create_agent()`.
2. **Layer your guardrails** — multiple safety layers provide better protection.
3. **Deterministic first, model-based second** — block simple cases before expensive LLM calls.
4. **HITL requires a checkpointer** — use `InMemorySaver` for development.
5. **Custom middleware** — use `before_agent()` and `after_agent()` for custom rules.

---

## 🧱 Layered Guardrails

```text
User Input
    ↓
Content Filter
    ↓
PII Protection
    ↓
Agent + Tools
    ↓
Human Approval
    ↓
Output PII Protection
    ↓
Safety Validation
    ↓
Safe Response

---

## 🏥 Real-World Application

The **Healthcare Chatbot** combines these concepts to:

- 🚫 Block harmful or off-topic requests
- 🔐 Protect sensitive patient information
- 👨‍⚕️ Require human approval for appointments
- ⚕️ Validate healthcare-related outputs

---

## 🎯 Final Takeaway

> **Guardrails are not a single safety check. They are multiple layers of protection around an AI agent.**

Using deterministic rules, PII protection, human oversight, and output validation together can help build more **controlled and reliable AI applications**.

---

## 📚 Additional Resources

- [LangChain Guardrails Docs](https://docs.langchain.com/oss/python/langchain/guardrails)
- [Middleware Docs](https://docs.langchain.com/oss/python/langchain/middleware/overview)
- [Human-in-the-Loop Docs](https://docs.langchain.com/oss/python/langchain/human-in-the-loop)
- [LangSmith for Observability](https://docs.langchain.com/oss/python/langchain/observability)
