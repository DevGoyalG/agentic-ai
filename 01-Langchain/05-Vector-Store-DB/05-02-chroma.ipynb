# Chroma

Chroma is an open-source, lightweight vector database designed specifically for building AI applications with Large Language Models (LLMs). It is highly favored by developers for Retrieval-Augmented Generation (RAG) systems, semantic search, and rapid prototyping because it can be embedded directly into local applications. Chroma is licensed under Apache 2.0.

https://python.langchain.com/v0.2/docs/integrations/vectorstores/

## Create Chroma Vector Store

### Import Required Libraries

```python
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

### Load the Document

```python
loader = TextLoader("speech.txt")
data = loader.load()
data
```

### Split the Document into Chunks

```python
### Split the Document into Chunks

text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=0)
splits = text_splitter.split_documents(data)
```

### Create Ollama Embeddings and Chroma Vector Database

```python
embedding = OllamaEmbeddings(model="mxbai-embed-large")
vectordb = Chroma.from_documents(documents = splits, embedding = embedding)
vectordb
```

## Querying

### Define the Query and Perform Similarity Search

```python
query = "What does the speaker believe is the main reason the United States should enter the war?"
docs = vectordb.similarity_search(query)
```

### View the Most Relevant Document

```python
docs[0].page_content
```

## Using Chroma as a Retriever

### Create a Retriever

```python
retriever=vectordb.as_retriever()
docs = retriever.invoke(query)
docs
```

### View Retrieved Document

```python
docs[0].page_content
```

## Similarity Search with Score

### Get Documents with Similarity Scores

```python
docs = vectordb.similarity_search_with_score(query)
docs
```

## Saving and Loading Chroma Vector Store

### Save Chroma Database to Disk

```python
### Saving to the disk

vectordb=Chroma.from_documents(documents=splits,embedding=embedding,persist_directory="./chroma_db")
```

### Load Chroma Database from Disk

```python
# Load from disk

db2 = Chroma(persist_directory="./chroma_db", embedding_function=embedding)
docs=db2.similarity_search(query)
print(docs[0].page_content)
```
