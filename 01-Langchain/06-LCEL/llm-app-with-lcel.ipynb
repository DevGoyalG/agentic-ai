# Build a Simple LLM Application with LCEL

This project demonstrates how to build a simple LLM application using LangChain Expression Language (LCEL). The application translates text from English into a specified language using a chat model, prompt templates, and output parsers.

In this application, you will learn how to:

- Use chat models with LangChain
- Create and use PromptTemplate / ChatPromptTemplate
- Process model responses with OutputParsers
- Build chains using LangChain Expression Language (LCEL)
- Connect prompts, models, and parsers into a complete LLM application
- Debugging and tracing your application using LangSmith
- Deploying your application with LangServe

## Environment and API Key Setup

### Loading Environment Variables

```python
import os 
from dotenv import load_dotenv
load_dotenv()
```

### Setting OpenAI API Key

```python
import openai
openai.api_key = os.getenv("OPEN_API_KEY")
```

### Loading Groq API Key

```python
groq_api_key = os.getenv("GROQ_API_KEY")
```

## Initialize the Chat Model

### Importing Chat Models and Creating the Groq Chat Model

```python
### Importing Chat Models
from langchain_openai import ChatOpenAI
from langchain_groq import ChatGroq

### Creating the Groq Chat Model
model = ChatGroq(model="openai/gpt-oss-120b", groq_api_key=groq_api_key)
model
```

## Working with Messages

### Creating System and Human Messages

```python
### Creating System and Human Messages
from langchain_core.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage(content="Translate the following from English to French"),
    HumanMessage(content="Hello, How are you?")
]

### Invoking the Chat Model
result = model.invoke(messages)
```

### Displaying the Model Response

```python
result
```

## Using Output Parsers

### Creating a String Output Parser and Parsing the Model Response

```python
### Creating a String Output Parser
from langchain_core.output_parsers import StrOutputParser
parser = StrOutputParser()

### Parsing the Model Response
parser.invoke(result)
```

## Using LCEL

### Creating an LCEL Chain and Invoking the LCEL Chain

```python
### Creating an LCEL Chain
chain = model|parser

### Invoking the LCEL Chain
chain.invoke(messages)
```

## Prompt Templates with LCEL

### Creating a Chat Prompt Template

```python
from langchain_core.prompts import ChatPromptTemplate

generic_template = "Translate the following into {language}:"

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", generic_template),
        ("user", "{text}")
    ]
)
```

### Invoking the Prompt Template

```python
result = prompt.invoke({"language": "French", "text": "Hello"})
```

### Displaying the Prompt

```python
result
```

### Converting Prompt to Messages

```python
result.to_messages()
```

## Chaining Components with LCEL

### Creating a Prompt-Model-Parser Chain and Invoking the Complete LCEL Chain

```python
### Creating a Prompt-Model-Parser Chain
chain = prompt|model|parser

### Invoking the Complete LCEL Chain
chain.invoke({"language": "French", "text": "Hello"})
```
